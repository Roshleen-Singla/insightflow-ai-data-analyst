```python
def clean_city_text(raw: str) -> str:
    """Basic normalization: upper, strip pin codes/digits, drop
    parenthetical notes, collapse punctuation/whitespace."""
    if not isinstance(raw, str):
        return ""
    text = raw.upper()
    text = re.sub(r"\(.*?\)", " ", text)      # drop "(...)" notes
    text = re.sub(r"\d+", " ", text)           # drop digits/PIN codes
    text = re.sub(r"[.\-/]", " ", text)        # dots/hyphens/slashes -> space
    text = re.sub(r"\s+", " ", text).strip()
    return text


def match_city(raw: str, cutoff: float = 0.90, min_len: int = 5) -> str:
    """Return a canonical city name for a raw ship-city value,
    or 'OTHER' if nothing close enough is found.

    cutoff is intentionally strict (0.90) because short Indian city
    names (e.g. FAIZABAD vs FIROZABAD) are easy to confuse at lower
    thresholds -- a stricter cutoff trades some recall for much
    higher precision, which matters more when this feeds reporting.
    """
    cleaned = clean_city_text(raw)
    if not cleaned:
        return "OTHER"

    # Try comma-separated parts first (either side might be the city)
    candidates = [p.strip() for p in cleaned.split(",") if p.strip()]
    if cleaned not in candidates:
        candidates.append(cleaned)

    # 1) exact / alias match on any candidate part (highest confidence)
    for part in candidates:
        if part in EXACT_LOOKUP:
            return EXACT_LOOKUP[part]

    # 2) fuzzy match, but only on individual whole words >= min_len chars
    #    (avoids matching short fragments like "T.Q." or "DT" to a city)
    best_match, best_score = None, 0.0
    for part in candidates:
        for word in part.split(" "):
            if len(word) < min_len:
                continue
            matches = difflib.get_close_matches(word, EXACT_LOOKUP.keys(), n=1, cutoff=cutoff)
            if matches:
                score = difflib.SequenceMatcher(None, word, matches[0]).ratio()
                if score > best_score:
                    best_match, best_score = EXACT_LOOKUP[matches[0]], score

    return best_match if best_match else "OTHER"
```

### business analysis

```python
# Which category generates the most Amount?

category_analysis = (
    df.groupby("category")
    .agg(
        total_sales=("sales_amount", "sum"),
    )
    .sort_values("total_sales", ascending=False)
)

print(category_analysis)
```

```python
# Which state generates the most Amount?
sales_by_state = (
    df.groupby("ship_state")["sales_amount"].sum().sort_values(ascending=False)
)

print(sales_by_state)
```

```python
# How does revenue change over time?
monthly_sales = (
    df
    .set_index("order_date")
    .resample("ME")["sales_amount"]
    .sum()
)

print(monthly_sales)
```

```python
top_cities = (
    df.groupby("ship_city")["sales_amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_cities)
```

```python
# Which categories have the most orders?
category_analysis = (
    df.groupby("category")
    .agg(
        total_amount=("sales_amount", "sum"),
        total_quantity=("quantity", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_values("total_amount", ascending=False)
)

print(category_analysis)
```

```python
# What's the average order value?
average_order_value = df.groupby("order_id")["sales_amount"].sum().mean()
print(f"Average order value: {average_order_value:.2f}")
```
